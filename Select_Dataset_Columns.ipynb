```python
# Task 18: Select Dataset Columns

## Objective
Select specific rows and columns from a dataset using Pandas
loc and iloc.

## Deliverables
- Examples using loc
- Examples using iloc
- Output showing selected records
- Explanation of label-based and position-based selection
```

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris

# Load the Iris dataset
iris = load_iris()

# Convert it into a Pandas DataFrame
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Add readable species names
df["species"] = iris.target_names[iris.target]

print("Iris dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
print("EXAMPLE 1: SELECT ROWS AND COLUMNS USING loc")

loc_result = df.loc[
    0:4,
    ["sepal length (cm)", "petal length (cm)", "species"]
]

display(loc_result)
```

```python
print("EXAMPLE 2: SELECT SPECIFIC ROW LABELS")

selected_rows = df.loc[
    [0, 10, 20],
    ["sepal length (cm)", "petal length (cm)", "species"]
]

display(selected_rows)
```

```python
print("EXAMPLE 3: SELECT SETOSA RECORDS")

setosa_records = df.loc[
    df["species"] == "setosa",
    ["sepal length (cm)", "petal length (cm)", "species"]
]

display(setosa_records.head())
```

```python
print("EXAMPLE 1: SELECT ROWS AND COLUMNS USING iloc")

iloc_result = df.iloc[0:5, [0, 2, 4]]

display(iloc_result)
```

```python
print("EXAMPLE 2: SELECT SPECIFIC ROW POSITIONS")

selected_positions = df.iloc[
    [0, 10, 20],
    [0, 2, 4]
]

display(selected_positions)
```

```python
print("EXAMPLE 3: SELECT A BLOCK OF DATA")

selected_block = df.iloc[5:10, 0:4]

display(selected_block)
```

```python
# Select one column by its label
print("SINGLE COLUMN USING loc")
display(df.loc[:, "species"].head())

# Select multiple columns by their labels
print("MULTIPLE COLUMNS USING loc")
display(
    df.loc[
        :,
        ["sepal length (cm)", "sepal width (cm)", "species"]
    ].head()
)

# Select one column by its integer position
print("SINGLE COLUMN USING iloc")
display(df.iloc[:, 4].head())
```

```python
## Difference Between loc and iloc

- loc selects rows and columns using labels.
- iloc selects rows and columns using integer positions.
- loc includes the ending label in a slice.
- iloc excludes the ending integer position in a slice.
- Both can select particular rows and columns together.
```

```python
comparison = pd.DataFrame({
    "Accessor": ["loc", "iloc"],
    "Selection method": [
        "Labels",
        "Integer positions"
    ],
    "Example": [
        'df.loc[0:4, ["species"]]',
        'df.iloc[0:5, [4]]'
    ]
})

display(comparison)
```

```python
print("DATASET DIMENSIONS")
print("Shape:", df.shape)
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nCOLUMN NAMES")
print(df.columns.tolist())

print("\nDATA TYPES")
print(df.dtypes)
```
